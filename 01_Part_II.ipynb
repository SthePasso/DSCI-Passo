# Flynn's Taxonomy Code Examples
#### 1. SISD (Single Instruction, Single Data)

Classic sequential code — one operation, one data point at a time.

### 2. SIMD (Single Instruction, Multiple Data)

One operation applied to many data points at once (vectorization).

#### 3. MISD (Multiple Instruction, Single Data)

Multiple different operations on the same data point (rare; imagine redundant checks).

4. MIMD (Multiple Instruction, Multiple Data)

Different tasks running simultaneously on different data (true parallel computing).

```
Quick Comparison
Type	Instruction	Data	    Example
SISD	One	        One	    Laptop adding numbers one-by-one
SIMD	One	        Many	    GPU processing 1000 pixels with same filter
MISD	Many	        One	    3 redundant safety checks on same sensor
MIMD	Many	        Many	    Ptolemy: 1024 cores doing different calculations

Ptolemy is MIMD — 1,024 cores, each running different code on different data chunks.
```
