# Flynn's Taxonomy Code Examples
#### 1. SISD (Single Instruction, Single Data)

Classic sequential code — one operation, one data point at a time.

````python
# Add 5 to each number sequentially
data = [1, 2, 3, 4, 5]
result = []

for x in data:
    result.append(x + 5)  # One instruction, one data point per loop

print(result)  # [6, 7, 8, 9, 10]
````

### 2. SIMD (Single Instruction, Multiple Data)

One operation applied to many data points at once (vectorization).

````python
a = [1,2,3,4,5]
b = [9,8,7,6,5]
c = []
for i in range(len(a)):
    c.append(a[i] + b[i])
````

````python
import numpy as np

# Same operation on all elements simultaneously
data = np.array([1, 2, 3, 4, 5])
result = data + 5  # One instruction, ALL data points

print(result)  # [6 7 8 9 10]
````

#### 3. MISD (Multiple Instruction, Single Data)

Multiple different operations on the same data point (rare; imagine redundant checks).

````python
# Different operations on one piece of data
value = 10

# Multiple different instructions on same value
doubled = value * 2      # Instruction 1
squared = value ** 2     # Instruction 2
halved = value / 2       # Instruction 3

print(f"Doubled: {doubled}, Squared: {squared}, Halved: {halved}")
# Doubled: 20, Squared: 100, Halved: 5.0
````

4. MIMD (Multiple Instruction, Multiple Data)

Different tasks running simultaneously on different data (true parallel computing).

````python
# Simulate 4 processors working on different data, different tasks
processors = 4
data = [10, 20, 30, 40]

print("=== MIMD in Action ===\n")

for i in range(processors):
    # Each processor runs a DIFFERENT instruction on DIFFERENT data
    if i == 0:
        result = data[i] * 2
        print(f"Processor {i}: Multiply {data[i]} by 2 → {result}")
    
    elif i == 1:
        result = data[i] ** 2
        print(f"Processor {i}: Square {data[i]} → {result}")
    
    elif i == 2:
        result = data[i] + 100
        print(f"Processor {i}: Add 100 to {data[i]} → {result}")
    
    else:
        result = data[i] // 5
        print(f"Processor {i}: Divide {data[i]} by 5 → {result}")

print("\n✓ All 4 processors ran different tasks (pseudo) simultaneously!")
````

```
Quick Comparison
Type	Instruction	Data	    Example
SISD	One	        One	    Laptop adding numbers one-by-one
SIMD	One	        Many	    GPU processing 1000 pixels with same filter
MISD	Many	        One	    3 redundant safety checks on same sensor
MIMD	Many	        Many	    Ptolemy: 1024 cores doing different calculations

Ptolemy is MIMD — 1,024 cores, each running different code on different data chunks.
```

````python
import pandas as pd

# ============================================
# 1. CREATE A SAMPLE DATAFRAME
# ============================================
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'Salary': [50000, 60000, 75000, 55000],
    'Department': ['HR', 'IT', 'Finance', 'IT']
})

print("Original DataFrame:")
print(df)
print("\n")


# ============================================
# 2. WRITE TO CSV
# ============================================
df.to_csv('employees.csv', index=False)
print("✓ Saved to employees.csv")


# ============================================
# 3. READ FROM CSV
# ============================================
df_from_csv = pd.read_csv('employees.csv')
print("\nRead from CSV:")
print(df_from_csv)
print("\n")


# ============================================
# 4. WRITE TO EXCEL
# ============================================
df.to_excel('employees.xlsx', index=False, sheet_name='Staff')
print("✓ Saved to employees.xlsx")


# ============================================
# 5. READ FROM EXCEL
# ============================================
df_from_excel = pd.read_excel('employees.xlsx', sheet_name='Staff')
print("\nRead from Excel:")
print(df_from_excel)
print("\n")


# ============================================
# 6. WRITE TO JSON
# ============================================
df.to_json('employees.json', orient='records', indent=2)
print("✓ Saved to employees.json")


# ============================================
# 7. READ FROM JSON
# ============================================
df_from_json = pd.read_json('employees.json')
print("\nRead from JSON:")
print(df_from_json)
print("\n")


# ============================================
# 8. WRITE TO PARQUET (Fast, compressed)
# ============================================
df.to_parquet('employees.parquet', index=False)
print("✓ Saved to employees.parquet")


# ============================================
# 9. READ FROM PARQUET
# ============================================
df_from_parquet = pd.read_parquet('employees.parquet')
print("\nRead from Parquet:")
print(df_from_parquet)
````
